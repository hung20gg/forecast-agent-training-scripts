```python
from transformers import AutoConfig, Qwen3_5ForConditionalGeneration, Qwen3_5ForCausalLM, AutoModel, AutoTokenizer
import torch
```

```python
repo="Qwen/Qwen3.5-2B"
local_dir = "./models/qwen3_5_2B"
```

```python
config = AutoConfig.from_pretrained(repo, trust_remote_code=True)
config.save_pretrained(local_dir)
```

```python
config = AutoConfig.from_pretrained(local_dir)
```

```python
from transformers import AutoProcessor, AutoTokenizer

processor = AutoProcessor.from_pretrained(repo)
tokenizer = AutoTokenizer.from_pretrained(repo)
```

```python
processor.save_pretrained(local_dir)
```

```python
tokenizer.save_pretrained(local_dir)
```

```python
model = Qwen3_5ForConditionalGeneration(config)
model = model.to(dtype=torch.bfloat16)
```

```python
true_model = Qwen3_5ForConditionalGeneration.from_pretrained(repo, trust_remote_code=True)
```

```python
model.model.language_model.load_state_dict(true_model.model.language_model.state_dict())
model.lm_head.load_state_dict(true_model.lm_head.state_dict())
```
